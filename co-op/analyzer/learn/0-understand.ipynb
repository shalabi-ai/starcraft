```python
import sc2reader
import pandas as pd

from replay.army_processor import ArmyProcessor
not_ephemerals = ["Roach", "Dragoon", "Scout", "Baneling", "HighTemplar", "HunterKiller",
                  "TrooperMengsk", "TrooperMengskImproved", "HotSRaptor",
                  "HellbatBlackOps", "HotSRaptor", "DehakaSwarmHost"]
army_units = []
for path in paths:
    replay1 = sc2reader.load_replay(path)
    processor = ArmyProcessor(replay1)
    units, unit_events = processor.process_replay()

    rows = []

    for unit in units.values():
        if unit.is_army:
            army_units.append(unit)

        if unit.death_frame is None:
            continue
        lifetime = (
                unit.death_frame
                - unit.birth_frame
        )
        rows.append({
            "unit_id": unit.unit_id,
            "unit_type": unit.current_type,
            "is_army": unit.is_army,
            "birth_frame": unit.birth_frame,
            "lifetime": lifetime,
        })

    df = pd.DataFrame(rows)
    roach_df = df[df["unit_type"] == "Roach"]


    pd1 =  roach_df["lifetime"].describe(
            percentiles=[0.5, 0.9]
        )
    pd1 = roach_df.sort_values("lifetime")
    print(pd1)

    df["birth_cluster_size"] = (
        df.groupby(["unit_type", "birth_frame"])[["unit_type", "birth_frame"]]
        .transform("size")
    )
    df["spawn_signature"] = (
            df["unit_type"]
            + "_"
            + df["birth_cluster_size"].astype(str)
    )
    df = (
        df.groupby("spawn_signature")
        .agg(
            instances=("unit_id", "count"),
            median_lifetime=("lifetime", "median"),
            p90_lifetime=("lifetime", lambda x: x.quantile(0.9)),
        )
    )
    print(df)
    #ephemeral = processor.ephemeral_units(units)
    #print(ephemeral)
    df = processor.temporary_units(units)
    df = df[df["unit_type"].isin(not_ephemerals)]
    print(path)
    print(df)


#  not_ephemerals = ["Roach", "Scout", "HunterKiller", "HunterKiller", "HybridDestroyer", "HybridReaver", "InfestorTerran"]
```
